```python
# %matplotlib qt
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

plt.style.use('dark_background')
plt.rcParams['figure.max_open_warning'] = 100
```

```python
def readScalarFeatureI(i, subj_name, cond_age, cond_rest, fnames, dir_features):
    '''
    Reads the value of i-th scalar feature of a given subject after
    matching experimental conditions from the stored CSV files in the
    `dir_features`.
    '''
    pattern_rest = subj_name + '_' + cond_age + '_' + cond_rest
    fnames_cond = [f for f in fnames if pattern_rest in f]
    F = []
    for fname in fnames_cond:
        fpath_cond = os.path.join(dir_features, fname)
        df = pd.read_csv(fpath_cond)
        F.append(float(df['Values'][i]))
    if len(F) == 2:
        ftr = np.nanmean(F)
    elif len(F) == 1:
        ftr = F[0]
    else:
        print('A total of %d features were found for %s.' % (len(F), subj_name))
        raise ValueError('Unexpected number of features found.')
    return ftr

def collectScalarFeaturesI(i_ftr, cond_age, cond_rest, dir_features):
    '''
    Collect i-th scalar features for all subjects after matching
    experimental conditions from the stored CSV files in the
    `dir_features`.
    '''
    # Find all CSV files containing spectral features
    fnames = [f for f in os.listdir(dir_features) if f[-4:] == '.csv']
    subj_names = np.unique([f[:4] for f in fnames])
    
    subj_ftrs_i = np.empty((len(subj_names),))
    for i_subj, subj_name in enumerate(subj_names):
        subj_ftrs_i[i_subj] = readScalarFeatureI(i_ftr, subj_name, cond_age, cond_rest,
                                                 fnames, dir_features)
    return subj_ftrs_i, subj_names
```

```python
# Given
compute_lt_rel_mode = 'rest1' # 'mean': Avg feature from rest1 and rest2; 'rest1': Use
# features only from rest1; 'rest2': Use features only from rest2
compute_fmlt_mode = 'rest1' # 'mean': Avg fearture from rest1 and rest2; 'rest1': Use
# features only from rest1; 'rest2': Use features only from rest2
dir_results = os.path.expanduser('~/research/results/wash-u/')
plot_scatter = True
n_ftrs = 50
scalar_ftrs = {
    'fft': [i for i in range(n_ftrs) if i not in range(3, 14)],
    'fft-fooof': [24, 25, 26, 39, 40],
    'hht': [i for i in range(n_ftrs) if i not in range(3, 14)],
}
# n_scalar_ftrs = len(scalar_ftrs['fft'])
age_dict = {'4': 'Age-18', '5': 'Age-20'}
```

# Short-term reliability

```python
# Loop over all conditions
for spec_mode in ['fft', 'fft-fooof', 'hht']:
    
    # Create an array of nans to store st_rel values
    st_reliability = np.empty((n_ftrs, 3)); st_reliability[:] = np.nan

    # Set features directory
    dir_features = os.path.expanduser('~/research/results/wash-u/features/' + spec_mode)
    
    # Check if directory to store reliability results exist; if not, create it
    dir_rlbt = os.path.join(dir_results, 'reliability', spec_mode)
    Path(dir_rlbt).mkdir(parents=True, exist_ok=True)
    
    for i_age, age_group in enumerate(age_dict):
        for i_ftr in scalar_ftrs[spec_mode]:
            # Get Feature-i for all subjects in conditions `rest1` and `rest2`
            rest1_ftrs_i, _ = collectScalarFeaturesI(i_ftr, age_group, 'rest1', dir_features)
            rest2_ftrs_i, _ = collectScalarFeaturesI(i_ftr, age_group, 'rest2', dir_features)

            # Evaluate reliability in terms of Pearson's corr coeff
            rho = np.corrcoef(rest1_ftrs_i, rest2_ftrs_i)
            st_reliability[i_ftr, i_age] = abs(rho[0, 1])
            print('Short-term reliability of Feature-%02d in age group-%s is %0.2f.' %
                  (i_ftr + 1, age_group, rho[0, 1]))

            # Draw scatter plot of features and save figure
            if plot_scatter:
                fpath_fig = os.path.join(dir_rlbt, 'shortterm_reliability_agegroup=%s_ftr-%02d.pdf' %
                                         (age_group, i_ftr + 1))
                print('Saving scatter plot at', fpath_fig)
                plt.figure()
                plt.scatter(rest1_ftrs_i, rest2_ftrs_i)
                plt.title('%s: Corr Coef = %0.2f (%s)' %
                          (spec_mode.upper(), rho[0, 1], age_dict[age_group]))
                plt.xlabel('Feature-%02d (rest1)' % (i_ftr + 1))
                plt.ylabel('Feature-%02d (rest2)' % (i_ftr + 1))
                plt.savefig(fpath_fig, bbox_inches='tight')
                plt.show()

    # Calculate mean reliability from both resting conditions
    st_reliability[:, 2] = np.mean(st_reliability[:, :2], axis=1)

    # Form a dataframe table of short-term reliability values and export in CSV format
    st_rel_table = pd.DataFrame(st_reliability, columns=['Age-18', 'Age-20', 'Mean'],
                                index=range(1, n_ftrs + 1))
    fpath_st_rel_table = os.path.join(dir_rlbt, 'shortterm_reliability_values.csv')
    st_rel_table.to_csv(fpath_st_rel_table)
```

# Long-term reliability

```python
# Loop over all conditions
lt_reliability = np.empty((n_ftrs,)); lt_reliability[:] = np.nan
for spec_mode in ['fft', 'hht']:
    # Set features directory
    dir_features = os.path.expanduser('~/research/results/wash-u/features/' + spec_mode)
    
    # Check if directory to store reliability results exist; if not, create it
    dir_rlbt = os.path.join(dir_results, 'reliability', spec_mode)
    Path(dir_rlbt).mkdir(parents=True, exist_ok=True)
    
    for i_ftr in scalar_ftrs[spec_mode]:
        age_ftrs_i = []
        for age_group in age_dict:
            # Get Feature-i for all subjects in condition `rest1` and `rest2` and average them
            rest1_ftrs_i, _ = collectScalarFeaturesI(i_ftr, age_group, 'rest1', dir_features)
            rest2_ftrs_i, _ = collectScalarFeaturesI(i_ftr, age_group, 'rest2', dir_features)
#             age_ftrs_i.append((rest1_ftrs_i + rest2_ftrs_i) / 2)
            if compute_lt_rel_mode == 'mean':
                age_ftrs_i.append((rest1_ftrs_i + rest2_ftrs_i) / 2)
            elif compute_lt_rel_mode == 'rest1':
                age_ftrs_i.append(rest1_ftrs_i)
            elif compute_lt_rel_mode == 'rest2':
                age_ftrs_i.append(rest2_ftrs_i)
            else:
                print('compute_lt_rel_mode =', compute_lt_rel_mode)
                raise ValueError("Long-term reliability can only be computed in three modes:" +
                                 "'rest1', 'rest2', and 'mean'.")

        # Evaluate reliability in terms of Pearson's corr coeff
        rho = np.corrcoef(age_ftrs_i[0], age_ftrs_i[1])
        lt_reliability[i_ftr,] = abs(rho[0, 1])
        print('Long-term reliability of Feature-%02d is %0.2f.' % (i_ftr + 1, rho[0, 1]))

        # Draw scatter plot of features and save figure
        if plot_scatter:
            fpath_fig = os.path.join(dir_rlbt,
                                     'longterm_reliability_compmode=%s_ftr-%02d.pdf' %
                                     (compute_lt_rel_mode, i_ftr+1))
            print('Saving scatter plot at', fpath_fig)
            plt.figure()
            plt.scatter(age_ftrs_i[0], age_ftrs_i[1])
            plt.title('%s: Corr Coef = %0.2f' % (spec_mode.upper(), rho[0, 1]))
            plt.xlabel('Feature-%02d (Age-18)' % (i_ftr + 1))
            plt.ylabel('Feature-%02d (Age-21)' % (i_ftr + 1))
            plt.savefig(fpath_fig, bbox_inches='tight')
            plt.show()

    # Form a dataframe table of short-term reliability values and export in CSV format
    lt_rel_table = pd.DataFrame(lt_reliability, columns=['Age-18Vs21'], index=range(1, n_ftrs + 1))
    fpath_lt_rel_table = os.path.join(dir_rlbt, 'longterm_reliability_values_compmode=%s.csv'
                                      % compute_lt_rel_mode)
    lt_rel_table.to_csv(fpath_lt_rel_table)
```

# Familiality

```python
# Loop over all conditions
fmlt = np.empty((n_ftrs, 3)); fmlt[:] = np.nan
for spec_mode in ['fft', 'hht']:
    # Set features directory
    dir_features = os.path.expanduser('~/research/results/wash-u/features/' + spec_mode)
    
    # Check if directory to store familiality results exist; if not, create it
    dir_fmlt = os.path.join(dir_results, 'familiality', spec_mode)
    Path(dir_fmlt).mkdir(parents=True, exist_ok=True)
    
    for i_age, age_group in enumerate(age_dict):
        for i_ftr in scalar_ftrs[spec_mode]:
            # Get Feature-i for all subjects in conditions `rest1` and `rest2` and average them
            rest1_ftrs_i, subj_names = collectScalarFeaturesI(i_ftr, age_group, 'rest1', dir_features)
            rest2_ftrs_i, _ = collectScalarFeaturesI(i_ftr, age_group, 'rest2', dir_features)
            if compute_fmlt_mode == 'mean':
                age_ftrs_i = (rest1_ftrs_i + rest2_ftrs_i) / 2
            elif compute_fmlt_mode == 'rest1':
                age_ftrs_i = rest1_ftrs_i
            elif compute_fmlt_mode == 'rest2':
                age_ftrs_i = rest2_ftrs_i
            else:
                print('compute_fmlt_mode =', compute_fmlt_mode)
                raise ValueError("Familiality can only be computed in three modes:" +
                                 "'rest1', 'rest2', and 'mean'.")

            # Split age-specific features in 2 to compare between twins
            twinsA_age_ftrs_i = [age_ftrs_i[i_subj] for i_subj in range(0, len(subj_names), 2)]
            twinsB_age_ftrs_i = [age_ftrs_i[i_subj] for i_subj in range(1, len(subj_names), 2)]

            # Test if twin subjects are identified correctly
            twins = [(subj_names[i], subj_names[i + 1]) for i in range(0, len(subj_names), 2)]
            assert all([(int(b) - int(a)) == 1 for a, b in twins])

            # Evaluate familiality in terms of Pearson's corr coeff
            rho = np.corrcoef(twinsA_age_ftrs_i, twinsB_age_ftrs_i)
            fmlt[i_ftr, i_age] = abs(rho[0, 1])
            print('Familiality of Feature-%02d is %0.2f.' % (i_ftr + 1, rho[0, 1]))

            # Draw scatter plot of features and save figure
            if plot_scatter:
                fpath_fig = os.path.join(dir_fmlt, 'familiality_agegroup=%s_compmode=%s_ftr-%02d.pdf' %
                                         (age_group, compute_fmlt_mode, i_ftr + 1))
                print('Saving scatter plot at', fpath_fig)
                plt.figure()
                plt.scatter(twinsA_age_ftrs_i, twinsB_age_ftrs_i)
                plt.title('%s: Corr Coef = %0.2f (%s)' %
                          (spec_mode.upper(), rho[0, 1], age_dict[age_group]))
                plt.xlabel('Feature-%02d (Twin-A)' % (i_ftr + 1))
                plt.ylabel('Feature-%02d (Twin-B)' % (i_ftr + 1))
                plt.savefig(fpath_fig, bbox_inches='tight')
                plt.show()

    # Calculate mean familiality from both age groups
    fmlt[:, 2] = np.mean(fmlt[:, :2], axis=1)

    # Form a dataframe table of familiality values and export in CSV format
    fmlt_table = pd.DataFrame(fmlt, columns=['Age-18', 'Age-21', 'Mean'],
                              index=range(1, n_ftrs + 1))
    fpath_fmlt_table = os.path.join(dir_fmlt, 'familiality_values_compmode=%s.csv' % compute_fmlt_mode)
    fmlt_table.to_csv(fpath_fmlt_table)
```
