### Pfefferbaum et al. 1991

Channel(s): 15 electrodes (F7, F3, Fz, F4, F8, T3, C3, Cz, C4, T4, T5, P3, Pz, P4, T6); sternovertebral reference.

Filtering: 0.13-70 Hz (down 3 dB)

Fs: 200 Hz

Baseline: 100 ms

Epoch: 1150 ms

- Exclude trials with incorrect behavioral response, electrical artifacts (EEG >+- 200 μV), V/HEOG artifacts (>+- 100 μV).
- Rare stim trials averaged separately to give conventional ERP.
- P3 measured at Fz, Pz, and Cz between 275-600 ms.

### Glenn et al. 1993
Channel(s): Pz

Bandpass filtering: 0.3 - 100.0 Hz (analog)

Fs: 500 Hz

Baseline: 500 ms

Epoch: 500 ms

- If sampled EEG during baseline/epoch exceeds 75 mV peak-to-peak, discard trial
- Average trials for stim condition (attended target trials) and apply peak detection algorithm
- P3 amplitudes were absolute voltages measured from the prestim baseline

### Costa et al. 2000

Channel(s): 21 electrodes (10-20 system); referenced to nose.

Filtering: 0.02 - 50 Hz bandpass (online); 32 Hz lowpass (offline; digital; done on single-trial data)

Fs: 256 Hz

Baseline: 187 ms

Epoch: 1313 ms

- Artifact rejection: V/HEOG artifact rejection (eye movement AND EEG threshold = 73.3 μV p-p).
- P300 defined at each electrode as the largest positive-peak within a latency window of 275-575 ms.
- Peak amp was measured relative to prestim baseline, and peak latency was measured from the time of stim onset.
- P300 amp was averaged across the electrodes within anterior/posterior "factor"
- PCA was used to find factors: anterior (F3, F4, F7, F8, Fz, C3, C4, Cz); posterior (P3, P4, Pz, P7, P8, O1, O2)

### Hada et al. 2000

Channel(s): 61 electrodes

Filtering: 0.02 - 50 Hz bandpass (analog); 16 Hz lowpass (digital)

Fs: 256 Hz

Baseline: 187.5 ms

Epoch: 1323 ms

- Artifact rejection: EMG, EOG, saturation > 73.3 μV
- P3a and P3b components were defined as the largest positive-peak within a latency window of 250-450 ms.
- Peak amp was measured relative to prestim baseline, and peak latency was measured from the time of stim onset.

### Suresh et al. 2003

Channel(s): 61 electrodes; reference on nose.

Filtering: 0.02-50 Hz (online); 16 Hz low-pass (offline; digital)

Fs: 256 Hz

Baseline: 187.5 ms

Epoch: 1500 ms

- online artifact rejection: EMG, V/HEOG, and saturation artifact >= 73.3 μV
- only correct response trials without eye artifacts were averaged to get ERPs
- derived from the target stimulus in the rare condition
- P3 = largest peak in the latency window (250-450 ms; derived as +-100 ms from peak of grand averaged waveform)
- peak amp measured relative to prestim baseline; latency measured from stim onset

### Wan et al. 2010

Channel(s): 64 electrodes; referenced to linked mastoids (M1 and M2) .

Filtering: 0.1-35 Hz (offline; 12 dB/oct slope; continuous EEG for each channel)

Fs: 1000 Hz

Baseline: 100 ms

Epoch: 900 ms

- ERP epochs with eye blink amp >50 μV were rejected
- ERP epochs were then averaged for 2 conds: target and non-target
- All accepted trials, including trials without responses, were averaged
- Temporal and occipital channels discarded
- 9 channels selected in groups: F3, Fz, F4 (frontal); C3, Cz, C4 (central); P3, Pz, P4 (parietal)
- P3 detection: semi-automated alg searched for most postive peak over 200-600 ms; peaks then confirmed/adjusted manually

### Anderson et al. 2011

Channel(s): 64 electrodes (10-20 system); referenced to mastoids.

Filtering: 0.1 - 35 Hz (offline; 12 dB/oct slope)

Fs: 1000 Hz

Baseline: 100 ms

Epoch: 900 ms

- baseline correction; averaging trials within subjects
- V/HEOG-based artifact removal using spatial filter created for each subject -> manual deletion of contaminated epochs
- P3s defined as most positive voltage between 200-600 ms; averaged at 9 electrodes (F3, Fz, F4, C3, Cz, C4, P3, Pz, P4)
